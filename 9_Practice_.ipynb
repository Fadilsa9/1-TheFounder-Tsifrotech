```python
# ✅ تثبيت مكتبة scikit-learn (مكتبة تعلم الآلة الكلاسيكية)
# نستخدم هذه المكتبة لأنها تحتوي على خوارزميات جاهزة مثل الانحدار، الشجرة، SVM...
!pip install scikit-learn

# ✅ استيراد المكتبات الضرورية
import numpy as np  # للتعامل مع المصفوفات والأرقام
import matplotlib.pyplot as plt  # لعرض الرسومات البيانية لاحقاً (اختياري)
from sklearn.linear_model import LogisticRegression  # استيراد خوارزمية الانحدار اللوجستي من مكتبة scikit-learn

# ✅ إعداد البيانات الطبية (أعراض المرضى مشفّرة بأرقام، مع النتيجة)
# كل صف يمثل مريضًا، والأرقام تمثل الخصائص/الأعراض
X = np.array([
    [1, 2],   # مريض 1: لديه الأعراض 1 و2
    [2, 3],   # مريض 2: لديه الأعراض 2 و3
    [3, 4],   # مريض 3: لديه الأعراض 3 و4
    [4, 5],   # مريض 4: لديه الأعراض 4 و5
    [5, 6]    # مريض 5: لديه الأعراض 5 و6
])

# القيم المستهدفة: 0 = سليم، 1 = مريض
y = np.array([
    0,  # المريض الأول سليم
    0,  # المريض الثاني سليم
    0,  # المريض الثالث سليم
    1,  # المريض الرابع مريض
    1   # المريض الخامس مريض
])

# ✅ إنشاء النموذج (Object) من خوارزمية Logistic Regression
model = LogisticRegression()

# ✅ تدريب النموذج على البيانات
model.fit(X, y)
# هذا السطر يجعل النموذج يتعلّم العلاقة بين الأعراض (X) والنتيجة (y)

# ✅ اختبار النموذج على مريض جديد
# هذا المريض لديه أعراض: 6 و7
new_observation = np.array([[6, 7]])

# ✅ التنبؤ النهائي: هل المريض مريض (1) أم سليم (0)؟
prediction = model.predict(new_observation)

# ✅ طباعة النتيجة
print("🔮 التنبؤ النهائي:", prediction)

# ✅ طباعة احتمالات كل فئة (سليم أو مريض)
# يعرض احتمالية انتماء العينة لكل صنف (0 أو 1)
probabilities = model.predict_proba(new_observation)
print("📊 الاحتمالات لكل فئة [سليم، مريض]:", probabilities)
```

```python
# ✅ 1. تثبيت مكتبة scikit-learn (إذا لم تكن مثبتة بالفعل)
# مكتبة تعلم آلة جاهزة تحتوي على خوارزميات التصنيف، التجميع، الانحدار...
!pip install scikit-learn

# ✅ 2. استيراد المكتبات الضرورية
import numpy as np  # للتعامل مع المصفوفات
import matplotlib.pyplot as plt  # لإنشاء الرسومات البيانية
from sklearn.linear_model import LogisticRegression  # استيراد خوارزمية الانحدار اللوجستي

# ✅ 3. إعداد بيانات التدريب (رموز أعراض المرضى + النتيجة)
# كل صف في X يمثل مريضًا، وكل عمود يمثل أحد الأعراض
X = np.array([
    [1, 2],   # المريض الأول: سليم
    [2, 3],   # المريض الثاني: سليم
    [3, 4],   # المريض الثالث: سليم
    [4, 5],   # المريض الرابع: مريض
    [5, 6]    # المريض الخامس: مريض
])

# y: النتيجة المتوقعة لكل مريض
# 0 = سليم، 1 = مريض
y = np.array([0, 0, 0, 1, 1])

# ✅ 4. إنشاء النموذج وتدريبه على البيانات
model = LogisticRegression()     # إنشاء نموذج الانحدار اللوجستي
model.fit(X, y)                  # تدريب النموذج على البيانات

# ✅ 5. تجربة التنبؤ على مريض جديد
new_observation = np.array([[6, 7]])      # مريض جديد بأعراض جديدة
prediction = model.predict(new_observation)  # التنبؤ النهائي: هل مريض أم لا؟
print("🔮 التنبؤ:", prediction)              # الناتج: 1 ⇒ مريض

# ✅ 6. الحصول على الاحتمالات التفصيلية (احتمال أن يكون سليم أو مريض)
probabilities = model.predict_proba(new_observation)
print("📊 الاحتمالات [سليم، مريض]:", probabilities)



# ✅ 7. رسم خط الحدود بين الفئتين (Decision Boundary)

# نحدد حدود الرسم بناءً على مدى انتشار البيانات
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1  # حدود المحور X
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1  # حدود المحور Y

# نُنشئ شبكة من النقاط تغطي هذه الحدود
xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 0.01),  # القيم على المحور X
    np.arange(y_min, y_max, 0.01)   # القيم على المحور Y
)

# ندمج كل (x,y) في شبكة ونطلب من النموذج التنبؤ لكل نقطة
grid_points = np.c_[xx.ravel(), yy.ravel()]  # دمج المحاور في مصفوفة واحدة
Z = model.predict(grid_points)               # التنبؤ بكل نقطة
Z = Z.reshape(xx.shape)                      # إعادة تشكيل النتائج لتتناسب مع الرسم

# ✅ 8. رسم الخلفية (خريطة القرار)
plt.contourf(xx, yy, Z, alpha=0.8)           # تلوين المناطق حسب التصنيف (مريض/سليم)

# ✅ 9. رسم بيانات المرضى فوق الخلفية
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k')  # المرضى الحقيقيون

# ✅ 10. تخصيص الرسم
plt.xlabel("Основной симптом")             # تسمية المحور X
plt.ylabel("Сопутствующий симптом")        # تسمية المحور Y
plt.title("Логистическая регрессия")       # عنوان الرسم
plt.xlim(xx.min(), xx.max())               # تحديد حدود X
plt.ylim(yy.min(), yy.max())               # تحديد حدود Y

plt.show()  # عرض الشكل النهائي
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm  # استيراد مكتبة SVM من sklearn

# إنشاء بيانات وهمية تمثل أعراض مرضى
X = np.array([[1, 2], [3, 7], [3, 4], [4, 5], [5, 6]])  # كل صف يمثل مريضًا وله عرضان
y = np.array([0, 2, 0, 1, 1])  # التصنيفات (أنواع الأمراض): 0 أو 1 أو 2

h = 0.01  # خطوة الشبكة: كل كم نقطة نرسم فاصل قرار
C = 1.0   # معامل تنظيم النموذج لتقليل التعقيد أو الإفراط في التخصيص

# إنشاء نموذج SVM باستخدام نواة خطية (linear kernel)
svc = svm.SVC(kernel='linear', C=C, gamma=0).fit(X, y)

# تحديد حدود الرسم البياني لإظهار التصنيف بوضوح
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1  # حدود محور X
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1  # حدود محور Y

# إنشاء شبكة من النقاط ضمن حدود الرسم
xx, yy = np.meshgrid(
    np.arange(x_min, x_max, h),  # القيم عبر محور X بتدرج h
    np.arange(y_min, y_max, h)   # القيم عبر محور Y بتدرج h
)

# التنبؤ بتصنيف كل نقطة في الشبكة باستخدام النموذج المدرب
Z = svc.predict(np.c_[xx.ravel(), yy.ravel()])  # دمج كل x و y في صفوف لنمررها للنموذج
Z = Z.reshape(xx.shape)  # إعادة تشكيل Z لتتناسب مع شكل الشبكة (للرسم)

# رسم خريطة القرار (تلوين الخلفية حسب الفئة المتوقعة)
plt.contourf(xx, yy, Z, cmap=plt.cm.Paired, alpha=0.8)

# رسم بيانات المرضى الأصلية بلون الفئة
result = plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.Paired)

# توليد مفتاح الألوان لتوضيح كل فئة
plt.legend(result.legend_elements()[0], ['0', '1', '2'])

# تخصيص عناصر الرسم
plt.xlabel('Основной симптом')  # تسمية المحور X
plt.ylabel('Сопутствующий симптом')  # تسمية المحور Y
plt.xlim(xx.min(), xx.max())  # تعيين حدود محور X
plt.title('Метод опорных векторов с линейным ядром')  # عنوان الرسم

# عرض الرسم النهائي
plt.show()
```

```python
import numpy as np  # مكتبة العمليات العددية

# دالة لتدريب البرسيبترون باستخدام قاعدة روزنبلات
def perceptron_train(X, y, learning_rate=0.01, max_iter=1000):
    # X: مصفوفة الإدخالات (الخصائص)
    # y: مصفوفة المخرجات (الفئات المستهدفة)
    # learning_rate: معدل التعلم
    # max_iter: عدد التكرارات (الجولات التدريبية)

    m, n = X.shape  # عدد العينات m وعدد الخصائص n
    W = np.random.rand(n, 1)  # توليد أوزان عشوائية ابتدائية

    for _ in range(max_iter):  # نكرر عملية التدريب max_iter مرة
        y_pred = np.heaviside(np.dot(X, W), 0)  # توقع الفئة باستخدام دالة Heaviside (عتبة)
        error = y - y_pred  # الفرق بين التوقع والنتيجة الحقيقية
        delta_W = learning_rate * np.dot(X.T, error)  # تعديل الأوزان حسب الخطأ
        W += delta_W  # تحديث الأوزان

    return W  # إرجاع الأوزان النهائية

# مجموعة بيانات تمثل سجلات طبية، كل سجل له عرضان (Feature1, Feature2)
X = np.array([[1, 2], [3, 7], [3, 4], [4, 5], [5, 6]])

# الفئات المستهدفة: 0 أو 1 أو 2 (تشخيصات)
y = np.array([[0], [2], [0], [1], [1]])

# تدريب النموذج باستخدام الدالة السابقة
W = perceptron_train(X, y)

# طباعة الأوزان النهائية بعد التعلم
print('الأوزان النهائية للبرسيبترون:', W)
```

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

mu, sigma = 0, 0.1  # المتوسط والانحراف المعياري
s = np.random.normal(mu, sigma, 1000)  # إنشاء 1000 نقطة بتوزيع طبيعي

count, bins, ignored = plt.hist(s, 30, density=True)  # رسم التوزيع على شكل أعمدة (Histogram)

# رسم منحنى التوزيع الطبيعي (Probability Density Function)
plt.plot(bins, 1/(sigma * np.sqrt(2 * np.pi)) *
         np.exp( - (bins - mu)**2 / (2 * sigma**2) ),
         linewidth=2, color='r')
plt.show()
```

```python
# استيراد المكتبات اللازمة
import numpy as np                      # مكتبة لمعالجة المصفوفات
from matplotlib import pyplot as plt   # مكتبة للرسم البياني
import seaborn as sns                  # مكتبة لتحسين مظهر الرسومات
from sklearn.linear_model import LinearRegression  # استيراد الانحدار الخطي

# إنشاء بيانات الإدخال x والمخرجات y على شكل مصفوفات
X = np.array([[0.72], [0.45], [0.23], [0.76], [0.14]])  # X هو متغير الدخل (input)
y = np.array([0.54, 0.56, 0.38, 0.57, 0.17])            # y هو القيمة المستهدفة (target)

# إنشاء نموذج الانحدار الخطي
model = LinearRegression()       # إنشاء الكائن المسؤول عن التدريب
model.fit(X, y)                  # تدريب النموذج على البيانات

# رسم النقاط الأصلية (البيانات الحقيقية) باللون الأخضر
plt.scatter(X, y, color='g')

# رسم الخط الناتج من النموذج باللون الأسود
plt.plot(X, model.predict(X), color='k')
plt.show()                      # عرض الشكل النهائي

# توقع ناتج جديد عند X = 0.9
prediction = model.predict(np.array([[0.9]]))[0]
print("عند X = 0.9 فإن y =", prediction)

# عرض معامل الانحدار m (الميل)
print("m =", model.coef_[0])       # المعامل m يمثل الميل (slope)

# عرض الثابت c (التحوّل الرأسي)
print("c =", model.intercept_)     # الثابت c هو التقاطع مع محور y (intercept)
```

```python
import numpy as np
from matplotlib import pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# تعريف بيانات x و y
X = np.array([[0.72], [0.45], [0.23], [0.76], [0.14]])
y = np.array([0.54, 0.56, 0.38, 0.57, 0.17])

# تدريب نموذج الانحدار الخطي
model = LinearRegression()
model.fit(X, y)

# حساب القيم المتوقعة من النموذج المدرب
y_pred = model.predict(X)

# طباعة قيمة الخطأ المتوسط التربيعي MSE
print(mean_squared_error(y, y_pred))
print(f'\nالقيم الحقيقية:\n {y}')
print(f'قياس أداء النموذج، أي: إلى أي درجة كانت التنبؤات قريبة من القيم الحقيقية:\n {y_pred}')

# إعداد قيم مختلفة لـ w من -0.5 إلى 1
w = np.linspace(-0.5, 1, 100)
print(f'\n إعداد قيم مختلفة ل w: \n = {w}')


# رسم التأثير عند تثبيت w0 (الانحياز) وتغيير w1 (الميل)
y1 = np.outer(w, X) + model.intercept_
# print(f'\n y1: \n {y1}')

# رسم التأثير عند تثبيت w1 (الميل) وتغيير w0 (الانحياز)
y2 = np.outer(model.coef_[0] * np.ones(w.shape[0]), X) + np.array([w, w, w, w, w]).T

# دالة لحساب MSE بناءً على w1 و w0
def MSE(w1, w0):
    return mean_squared_error(y, w1 * X + w0)


# جعل الدالة قابلة للتطبيق على المصفوفات (تطبيق العنصر-عنصر)
vectorized_mse = np.vectorize(MSE)

# إنشاء شبكة من القيم w0 و w1 لحساب MSE لكل زوج ممكن
X_3d, Y_3d = np.meshgrid(w, w)
Z_3d = vectorized_mse(X_3d, Y_3d)

# بدء الرسم
fig = plt.figure(figsize=(16,5))
fig.suptitle('MSE')

# الرسم الأول: تأثير الميل w1
ax = fig.add_subplot(1, 3, 1)
ax.plot(w, [mean_squared_error(y,  np.array(y_w)) for y_w in y1], color='k')
ax.grid(True)
ax.set_ylabel('MSE')
ax.set_xlabel('w1 (наклон)')
ax.set_ylim(0, 1)

# الرسم الثاني: تأثير الانحياز w0
ax = fig.add_subplot(1, 3, 2)
ax.plot(w, [mean_squared_error(y,  np.array(y_w)) for y_w in y2], color='r')
ax.grid(True)
ax.set_ylabel('MSE')
ax.set_xlabel('w0 (отсечение)')
ax.set_ylim(0, 1)

# الرسم الثالث: رسم ثلاثي الأبعاد يوضح العلاقة بين w0 و w1 و MSE
ax = fig.add_subplot(1, 3, 3, projection='3d')
ax.plot_surface(X_3d, Y_3d, Z_3d)
ax.set_zlim(0, 1)
ax.set_zlabel('MSE')
ax.set_xlabel('w0 (отсечение)')
ax.set_ylabel('w1 (наклон)')

# عرض النتائج
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# إعداد العينة
np.random.seed(42)
n = 200

# 1️⃣ بيانات موزعة طبيعياً
X1 = np.random.normal(0, 1, n).reshape(-1, 1)
y1 = 3 * X1.squeeze() + np.random.normal(0, 0.5, n)  # علاقة خطية + ضوضاء طبيعية

# 2️⃣ بيانات موزعة غير طبيعياً (مثلاً: توزيع أسي)
X2 = np.random.exponential(scale=1.0, size=n).reshape(-1, 1)
y2 = 3 * X2.squeeze() + np.random.normal(0, 0.5, n)

# تدريب النموذج على كل مجموعة
model1 = LinearRegression().fit(X1, y1)
model2 = LinearRegression().fit(X2, y2)

# التنبؤ
y1_pred = model1.predict(X1)
y2_pred = model2.predict(X2)

# تقييم النموذجين
mse1 = mean_squared_error(y1, y1_pred)
mse2 = mean_squared_error(y2, y2_pred)

r2_1 = r2_score(y1, y1_pred)
r2_2 = r2_score(y2, y2_pred)

# الطباعة
print("🔹 نموذج البيانات الطبيعية:")
print(f"   - MSE: {mse1:.3f}")
print(f"   - R²: {r2_1:.3f}")

print("🔸 نموذج البيانات غير الطبيعية:")
print(f"   - MSE: {mse2:.3f}")
print(f"   - R²: {r2_2:.3f}")

# رسم النتائج
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# الرسم 1: البيانات الطبيعية
axs[0].scatter(X1, y1, label='بيانات فعلية')
axs[0].plot(X1, y1_pred, color='red', label='التنبؤ')
axs[0].set_title('بيانات موزعة طبيعياً')
axs[0].legend()

# الرسم 2: البيانات غير الطبيعية
axs[1].scatter(X2, y2, label='بيانات فعلية')
axs[1].plot(X2, y2_pred, color='red', label='التنبؤ')
axs[1].set_title('بيانات غير طبيعية')
axs[1].legend()

plt.tight_layout()
plt.show()
```
